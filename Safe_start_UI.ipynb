```python
%%html
<!DOCTYPE html>
<html lang="en">
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1.0">
<title>Safe Start Driving School Registration</title>

<style>
body {
    font-family: Arial, sans-serif;
    background: linear-gradient(135deg, #0f766e, #22c55e);
    display: flex;
    justify-content: center;
    align-items: center;
    height: 100vh;
    margin: 0;
}

.container {
    width: 100%;
    max-width: 450px;
}

.card {
    background: white;
    padding: 2rem;
    border-radius: 20px;
    box-shadow: 0 10px 25px rgba(0,0,0,0.2);
    text-align: center;
}

.logo {
    width: 120px;
    margin-bottom: 15px;
}

h2 {
    color: #065f46;
}

input,
select {
    width: 100%;
    margin: 10px 0;
    padding: 12px;
    border: 1px solid #ccc;
    border-radius: 8px;
    box-sizing: border-box;
}

button {
    width: 100%;
    margin-top: 15px;
    padding: 12px;
    border: none;
    border-radius: 8px;
    background: #065f46;
    color: white;
    font-size: 16px;
    cursor: pointer;
    font-weight: bold;
}

button:hover {
    background: #047857;
}

.back-btn {
    background: #d1d5db;
    color: black;
}

img {
    width: 150px;
    margin-top: 15px;
    border-radius: 10px;
    border: 1px solid #ddd;
}

.hidden {
    display: none;
}
</style>

</head>
<body>

<div class="container">

    <!-- STEP 1 -->
    <div id="step1" class="card">

        <!-- Replace with actual Safe Start logo -->
        <img src="https://via.placeholder.com/120x80?text=SAFE+START"
             class="logo">

        <h2>Safe Start Driving School</h2>
        <p>Student Registration Portal</p>

        <input type="text" id="name" placeholder="Full Name">

        <input type="email" id="email" placeholder="Email Address">

        <input type="tel" id="phone" placeholder="Phone Number">

        <select id="course">
            <option value="">Select Driving Class</option>
            <option>Class A (Motorcycle)</option>
            <option>Class B (Light Vehicles)</option>
            <option>Class C (Trucks)</option>
            <option>Class BCE (Commercial Vehicles)</option>
        </select>

        <button onclick="goToStep2()">
            Continue Registration
        </button>

    </div>


    <!-- STEP 2 -->
    <div id="step2" class="card hidden">

        <h2>Upload ID/Passport Photo</h2>

        <input
            type="file"
            id="studentImage"
            accept="image/*"
            onchange="previewImage(event)"
        >

        <img id="preview" class="hidden">

        <button onclick="submitRegistration()">
            Complete Registration
        </button>

        <button class="back-btn" onclick="goToStep1()">
            Back
        </button>

    </div>

</div>

<script>

function goToStep2() {

    const name = document.getElementById("name").value.trim();
    const email = document.getElementById("email").value.trim();
    const phone = document.getElementById("phone").value.trim();
    const course = document.getElementById("course").value;

    if (name && email && phone && course) {

        document.getElementById("step1")
            .classList.add("hidden");

        document.getElementById("step2")
            .classList.remove("hidden");

    } else {

        alert("Please fill in all fields.");

    }

}


function goToStep1() {

    document.getElementById("step1")
        .classList.remove("hidden");

    document.getElementById("step2")
        .classList.add("hidden");

    document.getElementById("preview")
        .classList.add("hidden");

    document.getElementById("studentImage").value = "";

}


function previewImage(event) {

    const image = document.getElementById("preview");

    image.src =
        URL.createObjectURL(event.target.files[0]);

    image.classList.remove("hidden");

}


function submitRegistration() {

    alert(
        "Registration completed successfully! Welcome to Safe Start Driving School."
    );

}

</script>

</body>
</html>
```
